```python
documents = [
    "Python is a programming language used for AI and automation.",
    "RAG stands for Retrieval-Augmented Generation.",
    "Embeddings represent text as numerical vectors.",
    "A vector database stores embeddings and helps retrieve relevant information.",
    "An LLM generates answers using the context provided to it."
]

print(documents)
```

## Load a pre-trained embedding model

```python
from sentence_transformers import SentenceTransformer


# "all-MiniLM-L6-v2" is a relatively small model designed to produce useful sentence embeddings
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
```

## Convert Documents into Embeddings

```python
document_embeddings = model.encode(documents)

print(document_embeddings.shape)
```

5 means we have five documents.

384 means the model represents each document using 384 numerical values.

## Search document by meaning

```python
query = "How does AI converts Text into Numbers?"
```

## Convert the question into embedding

```python
query_embedding = model.encode(query)

print(query_embedding.shape)
```

The shape (384,) means the question is represented by 384 numbers, just like each document.

## Compare the question with every document

```python
from sentence_transformers import util

similarities = util.cos_sim(query_embedding, document_embeddings)

print(similarities)
```

## Find the most relevant document

```python
# argmax - It finds the position of the highest similarity score.

best_match_index = similarities.argmax().item()

print(best_match_index)
```

```python
print("Query: ", query)
print("Best Match: ", documents[best_match_index])
print("Similarity Score: ", similarities[0][best_match_index].item())
```

## Top K-retrival

```python
# Find the indices of top 3 most similar documents

top_k = 3

scores = similarities[0]

top_indices = scores.argsort(descending=True)[:top_k]

print(top_indices)
print("=" * 80)

for idx in top_indices:
  index = idx.item()
  print("Document: ", documents[index])
  print("Similarity Score: ", scores[index].item())
  print("-"*80)


```

## Build Knowledge Base

```python

documents = [
    # Python and machine learning
    "Python is widely used in AI because it has libraries for data processing, machine learning, and deep learning. NumPy supports numerical computation, while pandas helps organize and analyze tabular data.",

    "Machine learning is a field of AI in which models learn patterns from data. Supervised learning uses labeled examples, while unsupervised learning discovers patterns in data without predefined labels.",

    "Neural networks contain layers of interconnected neurons. During training, the model adjusts its parameters to reduce a loss function that measures prediction errors.",

    # Embeddings
    "An embedding model converts text into a numerical vector. The vector represents learned features of the text, allowing computers to compare semantic similarity between pieces of text.",

    "Cosine similarity compares the direction of two vectors. A value closer to 1 indicates stronger directional alignment, while a value near 0 indicates little directional alignment. The interpretation depends on the embedding model and data.",

    # LLMs and prompting
    "A large language model, or LLM, is trained on large amounts of text to learn language patterns. It generates text by predicting tokens based on the preceding context.",

    "A prompt is the input given to an LLM. A prompt can contain instructions, a user question, examples, and supporting context. Clear instructions can help guide the model's response.",

    # RAG
    "Retrieval-Augmented Generation combines information retrieval with text generation. It retrieves relevant information from an external knowledge source and provides that information to an LLM as context.",

    "A typical RAG pipeline splits documents into chunks, creates an embedding for each chunk, stores the embeddings, embeds the user's question, retrieves relevant chunks, and passes them to an LLM.",

    "Chunking divides long documents into smaller sections. Chunks that are too large may contain unrelated information, while chunks that are too small may lose important context.",

    "A vector database stores vectors and can perform similarity searches to find vectors close to a query vector. Retrieved vectors are associated with their original text so the application can use that text as context.",

    "RAG can reduce reliance on an LLM's internal knowledge by supplying relevant external information. However, retrieval errors, incomplete documents, and model mistakes can still produce incorrect answers.",

    # Evaluation
    "RAG retrieval quality measures whether the system finds useful source passages. Answer quality measures whether the final response correctly and sufficiently answers the user's question using the available evidence."
]

print(f"Total documents: {len(documents)}")
```

```python
document_embeddings = model.encode(documents)

print("Embedding Matrix", document_embeddings.shape)
```

## Test with different questions

```python
queries = [
    "How does an AI model represent text as numbers?",
    "What is the difference between supervised and unsupervised learning?",
    "Why do we split documents into chunks in RAG?",
    "Can RAG guarantee that an LLM gives correct answers?"
]

for query in queries:
  query_embedding = model.encode(query)
  similarities = util.cos_sim(query_embedding, document_embeddings)
  score = similarities[0]
  top_results = score.argsort(descending=True)[:3]

  print("\n")
  print(f"Query: {query}")
  print("\n")
  print("-" * 80)

  for idx in top_results:
    index = idx.item()
    print(f"Document: {documents[index]}")
    print(f"Similarity Score: {score[index].item()}")
    print("-" * 80)
```

```python
query = "How does an AI model represent text as numbers?"

query_embedding = model.encode(query)
similarities = util.cos_sim(query_embedding, document_embeddings)[0]

ranked_indices = similarities.argsort(descending=True)

for rank, index in enumerate(ranked_indices[:5], start=1):
    i = index.item()
    print(f"Rank: {rank}")
    print(f"Score: {similarities[i].item():.4f}")
    print(f"Document: {documents[i]}")
    print("-" * 60)
```

```python
query = "How does an AI model represent text as numbers?"

query_embedding = model.encode(query)
similarities = util.cos_sim(query_embedding, document_embeddings)[0]

threshold = 0.40

relevant_indices = [
    i for i, score in enumerate(similarities)
    if score.item() >= threshold
]

if relevant_indices:
    relevant_indices.sort(
        key=lambda i: similarities[i].item(),
        reverse=True
    )

    for i in relevant_indices:
        print(f"Score: {similarities[i].item():.4f}")
        print(f"Document: {documents[i]}")
        print("-" * 50)
else:
    print("No sufficiently relevant documents found.")
```
